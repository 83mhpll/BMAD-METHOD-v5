**A Call Option**

```python
import numpy as np
class Call:
  def __init__(self,K,bid,ask,TTM):
    self.K = K
    self.bid = bid
    self.ask = ask
    self.TTM = TTM

  def get_payoff(self, prices):
        prices = np.array(prices)
        return np.maximum(prices - self.K, 0)
```

**Example:** Long Call, K = 5000



```python
import numpy as np
import matplotlib.pyplot as plt
prices = np.arange(4000, 6001, 10)

c = Call(5000,300,400,26122025)
```

```python
import numpy as np
import matplotlib.pyplot as plt
prices = np.arange(4000, 6001, 10)

c = Call(5000,300,400,26122025)
port = c.get_payoff(prices)

plt.figure(figsize=(8, 5))
plt.plot(prices, port)
plt.xlabel("Price")
plt.ylabel("Payoff")
plt.legend()
plt.grid(True)
plt.show()
```

**A Put Option**

```python
class Put:
    def __init__(self, K, bid, ask, TTM):
        self.K = K
        self.bid = bid
        self.ask = ask
        self.TTM = TTM

    def get_payoff(self, prices):
        prices = np.array(prices)
        return np.maximum(self.K - prices, 0)   # returns NumPy array
```

```python
import numpy as np
import matplotlib.pyplot as plt
prices = np.arange(4000, 6001, 10)

p = Put(5000,300,400,26122025)
port = p.get_payoff(prices)

plt.figure(figsize=(8, 5))
plt.plot(prices, port)
plt.xlabel("Price")
plt.ylabel("Payoff")
plt.legend()
plt.grid(True)
plt.show()
```

**Polymarket Bet**

```python
class Bet:
    def __init__(self, K, noRate, TTM):
        self.K = K
        self.noRate = noRate
        #noRate = 80 -> pay 80 today and get 100 if NO -> win 20
        self.TTM = TTM

    def get_payoff(self, prices):
        prices = np.array(prices)  # ensure NumPy array
        #np.where(condition, value_if_true, value_if_false)
        return np.where(prices < self.K, 100 - self.noRate, -self.noRate)
```

```python
b = Bet(5000,60,31122025)
result = b.get_payoff(4999.99)
print(result)
```

**Spot (Futures) Payoff**

```python
class Spot:
    def __init__(self, entry):
        self.entry = entry

    def get_payoff(self, prices):
        prices = np.array(prices)  # ensure NumPy array
        return prices - self.entry
```

```python
import numpy as np
import matplotlib.pyplot as plt
prices = np.arange(4000, 6001, 1)

s = Spot(5000)
port = s.get_payoff(prices)

plt.figure(figsize=(8, 5))
plt.plot(prices, port)
plt.xlabel("Price")
plt.ylabel("Payoff")
plt.legend()
plt.grid(True)
plt.show()
```
