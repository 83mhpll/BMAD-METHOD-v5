```python
import numpy as np
class Call:
  def __init__(self,K,bid,ask,TTM):
    self.K = K
    self.bid = bid
    self.ask = ask
    self.TTM = TTM

  def get_payoff(self, prices):
        prices = np.array(prices)
        return np.maximum(prices - self.K, 0)
```

```python
class Put:
    def __init__(self, K, bid, ask, TTM):
        self.K = K
        self.bid = bid
        self.ask = ask
        self.TTM = TTM

    def get_payoff(self, prices):
        prices = np.array(prices)
        return np.maximum(self.K - prices, 0)   # returns NumPy array
```

```python
class Bet:
    def __init__(self, K, noRate, TTM):
        self.K = K
        self.noRate = noRate
        self.TTM = TTM

    def get_payoff(self, prices):
        prices = np.array(prices)  # ensure NumPy array
        return np.where(prices < self.K, 100 - self.noRate, -self.noRate)
```

```python
class Spot:
    def __init__(self, entry):
        self.entry = entry

    def get_payoff(self, prices):
        prices = np.array(prices)  # ensure NumPy array
        return prices - self.entry
    def get_value(self,prices):
        value = prices
        return value
```

```python
class Bet2:
    def __init__(self, K, noRate, TTM):
        self.K = K
        self.noRate = noRate
        self.TTM = TTM

    def get_payoff(self, prices):
        prices = np.array(prices)  # ensure NumPy array
        return np.where(prices > self.K, 100 - self.noRate, -self.noRate)
```

```python
import numpy as np
import matplotlib.pyplot as plt
prices = np.arange(170, 450.1, 1)

p = Put(185,30.1,31.4,26122025)
b = Bet(400,82,31122025)
s = Spot(186.12)

port = s.get_payoff(prices) + (p.get_payoff(prices)-p.ask) + 1.8*b.get_payoff(prices)

plt.figure(figsize=(8, 5))
plt.plot(prices, port)
plt.xlabel("Price")
plt.ylabel("Payoff")
plt.legend()
plt.grid(True)
plt.show()
```
